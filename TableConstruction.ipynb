```python
#order: parks, visitor_use_stats, profile_summary, segments, segments_shares_monthly, spending_profiles, trip_characteristics
ct_parks = """
DROP TABLE IF EXISTS parks;
CREATE TABLE parks (
    park_code CHAR(4) PRIMARY KEY,
    park_name VARCHAR(100),
    state_id CHAR(2),
    N_bounding_coordinate DECIMAL(6,4),
    S_bounding_coordinate DECIMAL(6,4),
    E_bounding_coordinate DECIMAL(6,4),
    W_bounding_coordinate DECIMAL(6,4)
);
"""
```

```python
ct_visitor_use_stats = """
DROP TABLE IF EXISTS visitor_use_stats;
CREATE TABLE visitor_use_stats (
    park_code CHAR(4) PRIMARY KEY,
    month INT PRIMARY KEY,
    BC INT,
    CCG INT,
    CL INT,
    MISC INT,
    NROS INT,
    TH INT,
    TNRV INT,
    TNRVH INT,
    TOS INT,
    TRV INT,
    TRVH INT,
    TRVS INT,
    TT INT,
    TTRV INT,
    TV INT,
    FOREIGN KEY (park_code) REFERENCES parks (park_code)
        ON UPDATE RESTRICT
        ON DELETE RESTRICT
);
"""
```

```python
ct_profile_summary = """
DROP TABLE IF EXISTS profile_summary;
CREATE TABLE profile_summary (
    profile_name VARCHAR(100) PRIMARY KEY,
    park_code CHAR(4),
    coverage_months VARCHAR(100),
    FOREIGN KEY (park_code) REFERENCES parks (park_code)
        ON UPDATE RESTRICT
        ON DELETE RESTRICT
);
"""
```

```python
ct_segments = """
DROP TABLE IF EXISTS segments;
CREATE TABLE segments (
    segment VARCHAR(20) PRIMARY KEY,
    segment_description VARCHAR(255)
);
"""
```

```python
ct_segment_shares_monthly = """
DROP TABLE IF EXISTS segment_shares_monthly;
CREATE TABLE segment_shares_monthly (
    park_code CHAR(4) PRIMARY KEY,
    month INT PRIMARY KEY,
    segment VARCHAR(20) PRIMARY KEY,
    segment_split DECIMAL(5,4),
    FOREIGN KEY (park_code) REFERENCES parks (park_code)
        ON UPDATE RESTRICT
        ON DELETE RESTRICT,
    FOREIGN KEY (segment) REFERENCES segments (segment)
        ON UPDATE RESTRICT
        ON DELETE RESTRICT
);
"""
```

```python
ct_spending_profiles = """
DROP TABLE IF EXISTS spending_profiles;
CREATE TABLE spending_profiles (
    park_code CHAR(4) PRIMARY KEY,
    segment VARCHAR(20) PRIMARY KEY,
    profile_name VARCHAR(100),
    spending_category VARCHAR(100) PRIMARY KEY,
    spending_group VARCHAR(100),
    reporting_group VARCHAR(100),
    spending_pppdn DECIMAL(10,2),
    FOREIGN KEY (park_code) REFERENCES parks (park_code)
        ON UPDATE RESTRICT
        ON DELETE RESTRICT,
    FOREIGN KEY (segment) REFERENCES segments (segment)
        ON UPDATE RESTRICT
        ON DELETE RESTRICT,
    FOREIGN KEY (profile_name) REFERENCES profile_summary (profile_name)
        ON UPDATE RESTRICT
        ON DELETE RESTRICT
);
"""
```

```python
ct_trip_characteristics = """
DROP TABLE IF EXISTS trip_characteristics;
CREATE TABLE trip_characteristics (
    park_code CHAR(4) PRIMARY KEY,
    segment VARCHAR(20) PRIMARY KEY,
    profile_name VARCHAR(100),
    entry_rate DECIMAL(5,2),
    length_of_stay_attributable DECIMAL(5,2),
    length_of_stay_local_area DECIMAL(5,2),
    party_size DECIMAL(12,11),
    percent_nonlocal DECIMAL(12,11),
    trip_purpose_equal DECIMAL(12,11),
    trip_purpose_incidental DECIMAL(12,11),
    trip_purpose_primary DECIMAL(12,11),
    FOREIGN KEY (park_code) REFERENCES parks (park_code)
        ON UPDATE RESTRICT
        ON DELETE RESTRICT,
    FOREIGN KEY (segment) REFERENCES segments (segment)
        ON UPDATE RESTRICT
        ON DELETE RESTRICT,
    FOREIGN KEY (profile_name) REFERENCES profile_summary (profile_name)
        ON UPDATE RESTRICT
        ON DELETE RESTRICT
);
"""
```
