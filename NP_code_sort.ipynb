```python
#Code to sort by park code for the Visitor Spending Effect

import pandas as pd

# Read the CSV
data = pd.read_csv("Main_State_Data.csv")

# Read the Excel file containing the park codes
park_codes = pd.read_excel("NP_entries.xlsx")

# Keep only rows from the CSV where the Park Code
# exists in the Excel file
result = data[data["park_code"].isin(park_codes["Park Code"])]

# Display the result
print(result)

# Save the result
result.to_csv("Main_State_Data_NP.csv", index=False)
```

```python
#Code for sort by park code and year for Vistor Use Statitics

import pandas as pd

# Read the CSV
data = pd.read_csv("Main_State_Data.csv")

# Read the Excel file containing the park codes
park_codes = pd.read_excel("NP_entries.xlsx")

# Make sure Year is numeric
data["Year"] = pd.to_numeric(data["Year"], errors="coerce")

# Filter for National Park codes and 2024
result = data[
    (data["UnitCode"].isin(park_codes["Park Code"])) &
    (data["Year"] == 2024)
]

# Sort by UnitCode
result = result.sort_values(by="UnitCode")

# Display the result
print(result)

# Save the result
result.to_csv("Main_State_Data_NP_2024.csv", index=False)
```
