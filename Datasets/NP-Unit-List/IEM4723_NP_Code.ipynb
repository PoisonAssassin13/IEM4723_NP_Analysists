```python
from google.colab import files
uploaded = files.upload()
import pandas as pd

file_path = "NPS-Unit-List.xlsx"
df = pd.read_excel(file_path)

matches = df[
    df.astype(str).apply(
        lambda row: row.str.strip().str.upper().eq("NP").any(), axis=1
    )
]

print(matches)

matches.to_excel("NP_entries.xlsx", index=False)

print(f"Found {len(matches)} row contaning NP.")

files.download("NP_entries.xlsx")
```
